```python
from sklearn.cluster import DBSCAN
from sklearn.datasets import make_moons
import matplotlib.pyplot as plt
import numpy as np
```

```python
X, y = make_moons(n_samples=1000, noise=0.05, random_state=42)
dbscan = DBSCAN(eps=0.05, min_samples=5)
dbscan.fit(X)
```

```python
plt.figure(figsize=(8,5))
mask_nonoutliers = (dbscan.labels_ != -1)
mask_outliers = (dbscan.labels_ == -1)
X_without_outlier = X[mask_nonoutliers]
labels_without_outlier = dbscan.labels_[mask_nonoutliers]
X_with_outliers = X[mask_outliers]
plt.scatter(X_without_outlier[:, 0], X_without_outlier[:, 1], c=labels_without_outlier, cmap='viridis', s=10)
plt.scatter(X_with_outliers[:, 0], X_with_outliers[:, 1], s=20, marker='x', c='red')
```

```python
X, y = make_moons(n_samples=1000, noise=0.05, random_state=42)
dbscan = DBSCAN(eps=0.2, min_samples=5)
dbscan.fit(X)
```

```python
plt.figure(figsize=(8,5))
mask_nonoutliers = (dbscan.labels_ != -1)
mask_outliers = (dbscan.labels_ == -1)
X_without_outlier = X[mask_nonoutliers]
labels_without_outlier = dbscan.labels_[mask_nonoutliers]
X_with_outliers = X[mask_outliers]
plt.scatter(X_without_outlier[:, 0], X_without_outlier[:, 1], c=labels_without_outlier, cmap='viridis', s=10)
plt.scatter(X_with_outliers[:, 0], X_with_outliers[:, 1], s=20, marker='x', c='red')
```

DBSCAN only have fit_predict() not predict(), so use KNeighborsClassifier

```python
from sklearn.neighbors import KNeighborsClassifier
knn = KNeighborsClassifier(n_neighbors=50)
knn.fit(dbscan.components_, dbscan.labels_[dbscan.core_sample_indices_])

X_new = np.array([[-0.5, 0], [0, 0.5], [1,-0.1], [2, 1]])
print(knn.predict(X_new))
```
