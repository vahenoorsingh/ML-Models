```python
import torch
import torch.nn as nn
```

```python
class Dense(nn.Module):
  def __init__(self, in_features, out_features):
    super().__init__()
    self.linear = nn.Linear(in_features, out_features)
    self.relu = nn.ReLU()
  def forward(self, X):
    return self.relu(self.linear(X))
```

```python
torch.manual_seed(42)
dense = Dense(3, 5)
X = torch.randn(2, 3)
y_pred = dense(X)
y_pred.shape
```

```python
y_pred_check = dense.relu(X @ dense.linear.weight.T + dense.linear.bias)
print(torch.allclose(y_pred, y_pred_check))
```

```python
class Dense(nn.Module):
  def __init__(self, in_features, out_features):
    super().__init__()
    self.linear = nn.Parameter(torch.randn(out_features, in_features))
    self.relu = nn.ReLU()
  def forward(self, X):
    z = X @ self.weight.T + self.bias
    return self.relu(z)
```
