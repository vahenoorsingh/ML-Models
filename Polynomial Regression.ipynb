```python
import numpy as np
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
```

```python
rng = np.random.default_rng(seed=42)
m = 200 # number of instances
X = 6 * rng.random((m, 1)) - 3
y = 0.5 * X ** 2 + X + 2 + rng.standard_normal((m, 1))
```

```python
poly_features = PolynomialFeatures(degree=2, include_bias=False)

X_poly = poly_features.fit_transform(X)
```

```python
lin_reg = LinearRegression()
lin_reg.fit(X_poly, y)

```

```python
x_curve = np.linspace(np.min(X), np.max(X), 500)
y_curve = lin_reg.predict(poly_features.transform(x_curve.reshape(-1, 1)))
plt.scatter(X,y,s=10, color='blue', label='Data points')
plt.plot(x_curve,y_curve, color='red', linewidth=2, label='Polynomial Regression Curve')
plt.title('Polynomial Regression')
plt.xlabel('X')
plt.ylabel('y')
plt.grid()
plt.legend()
plt.show()
```
